```python
import numpy as np

# Inputs and target
x = np.array([1, 0, 1])
y = np.array([1])
lr = 0.001

# Weights initialization based on Laboratory Task 2
W_h = np.array([
    [ 0.2, -0.3],
    [ 0.4,  0.1],
    [-0.5,  0.2]
])

W_o = np.array([
    [-0.3],
    [-0.2]
])

theta_h = np.array([-0.4, 0.2])
theta_o = np.array([0.1])

# Activation function (ReLU) and its derivative
def relu(z):
    return np.maximum(0, z)

def relu_derivative(z):
    return np.where(z > 0, 1, 0)

# --- Forward Pass ---
Z_h = np.dot(x, W_h) + theta_h
H = relu(Z_h)

Z_o = np.dot(H, W_o) + theta_o
y_hat = relu(Z_o)

# Compute Error and Loss
error = y_hat - y
loss = 0.5 * np.sum(error ** 2)

print(f"Predicted Output (y_hat): {y_hat[0]}")
print(f"Error (y_hat - y): {error[0]}")
print(f"Loss: {loss}")

# --- Backward Pass ---
d_activation_o = relu_derivative(Z_o)
delta_o = error * d_activation_o

dW_o = np.outer(H, delta_o)
dtheta_o = delta_o

d_activation_h = relu_derivative(Z_h)
delta_h = np.dot(delta_o, W_o.T) * d_activation_h

dW_h = np.outer(x, delta_h)
dtheta_h = delta_h

# --- Parameter Update ---
W_o -= lr * dW_o
theta_o -= lr * dtheta_o
W_h -= lr * dW_h
theta_h -= lr * dtheta_h

print("\nUpdated Output Weights:\n", W_o)
print("Updated Hidden Weights:\n", W_h)
```
