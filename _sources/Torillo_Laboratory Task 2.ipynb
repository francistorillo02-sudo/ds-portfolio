## 1. Given Parameters & Inputs
* **Input ($x$):** $\begin{bmatrix} 1 \\ 0 \\ 1 \end{bmatrix}$
* **Actual Output ($y$):** $\begin{bmatrix} 1 \end{bmatrix}$
* **Activation Function:** $f(z) = \max(0, z)$ (ReLU)
* **Hidden Unit Weights ($W_h$):** 
  $$\begin{bmatrix} w_{11} = 0.2 & w_{12} = -0.3 \\ w_{13} = 0.4 & w_{14} = 0.1 \\ w_{15} = -0.5 & w_{16} = 0.2 \end{bmatrix}$$
* **Output Unit Weights ($W_o$):** 
  $$\begin{bmatrix} w_{21} = -0.3 \\ w_{22} = -0.2 \end{bmatrix}$$
* **Biases ($\theta$):**
  $$\theta_1 = -0.4, \quad \theta_2 = 0.2, \quad \theta_3 = 0.1$$

---

## 2. Forward Pass Calculation

* **Step 1: Compute Hidden Layer Weighted Sums ($Z$)**
  $$Z_1 = (w_{11} \cdot x_1) + (w_{13} \cdot x_2) + (w_{15} \cdot x_3) + \theta_1$$
  $$Z_1 = (0.2 \cdot 1) + (0.4 \cdot 0) + (-0.5 \cdot 1) + (-0.4) = 0.2 + 0 - 0.5 - 0.4 = -0.7$$

  $$Z_2 = (w_{12} \cdot x_1) + (w_{14} \cdot x_2) + (w_{16} \cdot x_3) + \theta_2$$
  $$Z_2 = (-0.3 \cdot 1) + (0.1 \cdot 0) + (0.2 \cdot 1) + 0.2 = -0.3 + 0 + 0.2 + 0.2 = 0.1$$

* **Step 2: Apply Activation Function ($H = \max(0, Z)$)**
  $$H_1 = \max(0, -0.7) = 0$$
  $$H_2 = \max(0, 0.1) = 0.1$$

* **Step 3: Compute Output Layer Weighted Sum ($Z_3$)**
  $$Z_3 = (w_{21} \cdot H_1) + (w_{22} \cdot H_2) + \theta_3$$
  $$Z_3 = (-0.3 \cdot 0) + (-0.2 \cdot 0.1) + 0.1 = 0 - 0.02 + 0.1 = 0.08$$

* **Step 4: Predicted Output ($\hat{y}$)**
  $$\hat{y} = \max(0, Z_3) = \max(0, 0.08) = 0.08$$

---

## 3. Error Computation
* **Raw Error ($\delta$):**
  $$\delta = \hat{y} - y = 0.08 - 1 = -0.92$$
* **Mean Squared Error Loss ($E$):**
  $$E = \frac{1}{2}(\hat{y} - y)^2 = \frac{1}{2}(-0.92)^2 = 0.4232$$
